```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from tqdm import tqdm
import time

# Базовый URL сайта
topic_url='https://lifehacker.ru/topics/technology'
base_url='https://lifehacker.ru'

# Список для хранения данных
data = []
parsed_url=[]

def soup(url):
    response = requests.get(url, timeout=60)
    return BeautifulSoup(response.text, 'lxml')

for page_num in range (1, 11):
    page_url=topic_url+'?page='+str(page_num)
    raw_links=soup(page_url).find_all('a', class_='lh-small-article-card__link')
    parsed_url+=[item.get('href') for item in raw_links]

print(parsed_url[1])
for article in tqdm(parsed_url):
    article_url=base_url+article
    title=soup(article_url).find('h1', class_='article-card__title').text
    content=soup(article_url).find(
            'article',
            class_='single-article__post-content single-article__content-container',
            ).text
    data.append({
        'title':title,
        'content':content    
    })
df=pd.DataFrame(data)
pd.set_option('display.max_colwidth', 400)
df.head(10)
```
